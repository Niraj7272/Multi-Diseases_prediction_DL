# Data Collection

## DeepDiseases: Multi-Disease Prediction Using Artificial Neural Network

This notebook documents the source, collection method, and initial description of the dataset used for the DeepDiseases project.

## 1. Dataset Source

The dataset used in this project is the **Disease-Symptom Dataset**, obtained from Kaggle.

- **Dataset Name:** Disease-Symptom Dataset
- **Source:** Kaggle
- **Provider:** Dhivyeshrk
- **Dataset URL:** https://www.kaggle.com/datasets/dhivyeshrk/diseases-and-symptoms-dataset
- **Download Date:** September 2026
- **Target Variable:** `Diseases`

## 2. Data Collection Method

The dataset was downloaded from Kaggle in CSV format and stored in the project's raw data directory.

The original dataset contains disease names along with their associated symptoms. Each symptom is represented as a feature, where the value indicates whether the symptom is present for a particular record.

The collected dataset was preserved in its raw form. Data loading, cleaning, duplicate handling, and other preprocessing operations were performed separately in:

`02_data_loading_cleaning.ipynb`

## 3. Dataset Description

The dataset contains disease-symptom records designed for disease prediction based on symptom patterns.

### Dataset Structure

- **Target:** `Disease`
- **Features:** Symptom columns
- **Feature Representation:** Binary values (`0` and `1`)
- **0:** Symptom not present
- **1:** Symptom present

The original dataset contains **246,945 records**, **377 symptom features**, and **773 disease categories**.
